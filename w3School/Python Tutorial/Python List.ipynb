```python
#condition
thislist = ["apple", "banana", "cherry", "orange", "kiwi", "melon", "mango"]
if "apple" in thislist:
    print("apple available")
```

```python
#change value
thislist = ["apple", "banana", "cherry", "orange", "kiwi", "melon", "mango"]
thislist[0] = "Grape"
print(thislist)
```

```python
thislist = ["apple", "banana", "cherry", "orange", "kiwi", "mango"]
thislist[1:3] = ["grape","lemon"]
print(thislist)
```

```python
#add value on the list without override the existed value
thislist = ["apple", "banana", "cherry", "orange", "kiwi", "mango"]
thislist.insert(2,"grape")
print(thislist)
```
