```python
#mean (Get Average Value base on array data)

import numpy as np

speed = [99,86,87,88,111,86,103,87,94,78,77,85,86]

print(np.mean(speed))
```

```python
#median (Get Centre Value base on array data

import numpy as np
speed = [99,86,87,88,111,86,103,87,94,78,77,85,86]

print(np.median(speed))
```

```python
#median (Get Centre Value base on array data. If there are two numbers in the middle, 
#divide the sum of those numbers by two. if no centre value,then (86 + 87) / 2 = 86.5

import numpy as np
speed = [99,86,87,88,86,103,87,94,78,77,85,86]
print(np.median(speed))
```

```python
#Mode (Get Repeat Value base on array data

from scipy import stats

speed = [99,86,87,88,111,86,103,87,94,78,77,85,86]

print(stats.mode(speed)

```
