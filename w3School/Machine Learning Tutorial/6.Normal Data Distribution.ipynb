```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

# Data: Simulasi skor peperiksaan pelajar (Normal Distribution)
mean_score = 70  # Purata markah
std_dev_score = 10  # Sisihan piawai
num_students = 1000  # Bilangan pelajar

# Hasilkan data markah berdasarkan taburan normal
scores = np.random.normal(mean_score, std_dev_score, num_students)

# Plot histogram untuk taburan normal
plt.figure(figsize=(10,5))
plt.hist(scores, bins=30, density=True, alpha=0.6, color='blue', edgecolor='black')

# Tambahkan garisan taburan normal
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p = stats.norm.pdf(x, mean_score, std_dev_score)
plt.plot(x, p, 'k', linewidth=2)

# Garisan Mean dan Standard Deviation
plt.axvline(mean_score, color='red', linestyle='dashed', linewidth=2, label="Mean (70)")
plt.axvline(mean_score + std_dev_score, color='green', linestyle='dashed', linewidth=2, label="Mean +1 SD (80)")
plt.axvline(mean_score - std_dev_score, color='green', linestyle='dashed', linewidth=2, label="Mean -1 SD (60)")

# Tajuk dan Label
plt.title('Taburan Normal: Skor Peperiksaan Pelajar')
plt.xlabel('Skor Peperiksaan')
plt.ylabel('Kepadatan Kebarangkalian')
plt.legend()
plt.grid(True)

# Tunjukkan plot
plt.show()
```
