```python
import numpy as np

# Data masa ketibaan pekerja dalam minit
masa_ketibaan = np.array([8, 9, 7, 10, 8])

# Kira Variance & Standard Deviation
variance = np.var(masa_ketibaan)
std_dev = np.std(masa_ketibaan)

print(f"Variance: {variance}")
print(f"Standard Deviation: {std_dev}")
```

```python
aduan = np.array([53, 53, 55, 44, 55, 55, 55])  # Data aduan harian
variance_aduan = np.var(aduan)
std_dev_aduan = np.std(aduan)

print(f"Variance: {variance_aduan}")
print(f"Standard Deviation: {std_dev_aduan}")

# Tentukan kategori
if variance_aduan > std_dev_aduan:
    print("Bilangan aduan tidak konsisten (Buruk)")
else:
    print("Bilangan aduan konsisten (Baik)")
```
