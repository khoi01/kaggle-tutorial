```python
#The first array will have the mean set to 5.0 with a standard deviation of 1.0.
#The second array will have the mean set to 10.0 with a standard deviation of 2.0

import numpy as np
import matplotlib.pyplot as pylt

x = np.random.normal(5.0, 1.0, 1000)
y = np.random.normal(10.0, 2.0, 1000)
plt.scatter(x, y)
plt.show()
```

```python
#Positive Correlation (⬆️⬆️) → As X increases, Y increases (e.g., experience vs. salary).
#Negative Correlation (⬆️⬇️) → As X increases, Y decreases (e.g., age vs. reaction time).
#No Correlation (Random dots) → No relationship (e.g., shoe size vs. intelligence).


import matplotlib.pyplot as plt

# Example: Ad Budget vs Sales Revenue
ad_budget = [1000, 2000, 3000, 4000, 5000]  # X-axis (in USD)
sales_revenue = [5000, 7000, 8000, 11000, 15000]  # Y-axis


# Create Scatter Plot
plt.scatter(ad_budget, sales_revenue)
plt.xlabel("Advertising Budget (USD)")
plt.ylabel("Sales Revenue (USD)")
plt.title("Ad Budget vs Sales Revenue")

# Show the graph
plt.show()
```
