```python
#What is the 75. percentile? The answer is 43, meaning that 75% of the people are 43 or younger.


import numpy as np

ages = [5,31,43,48,50,41,7,11,15,39,80,82,32,2,8,6,25,36,27,61,31]

print(np.percentile(ages,75))
```

```python
import numpy as np

# Data bilangan aduan awam dalam 10 hari
aduan = np.array([50, 52, 49, 51, 100, 102, 98, 53, 55, 50])

# Kira Percentile
p10 = np.percentile(aduan, 10)  # 10th percentile
p50 = np.percentile(aduan, 50)  # 50th percentile (Median)
p90 = np.percentile(aduan, 90)  # 90th percentile

# Klasifikasi kategori
kategori = []
for nilai in aduan:
    if nilai <= p50:
        kategori.append("Baik: "+str(nilai))  # Rendah
        print(nilai)
    elif nilai <= p90:
        kategori.append("Sederhana: "+str(nilai))
    else:
        kategori.append("Buruk: "+str(nilai))  # Tinggi

# Paparkan keputusan
for i, nilai in enumerate(aduan):
    print(f"Aduan {nilai}: {kategori[i]}")
```
