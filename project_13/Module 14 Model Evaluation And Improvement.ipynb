```python
from ultralytics import YOLO

model = YOLO("yolov8s.pt")
results = model.train(
    data="data.yaml",
    epochs=50,
    imgsz=640,
    batch=8,
    project="runs",
    name="ball-yolov8"
)

results_dict = results.results_dict
```

```python
#CREATE A LOG FOR EACH MODEL THAT JUST CREATED.
from ultralytics import YOLO
import pandas as pd
import os

# Example: After training or validation
model = YOLO("model/v1_ball.pt")  # Replace with your model
results = model.val()               # or .train(...)

# Extract result metrics
summary = {
    "Version": "v1",
    "mAP50": float(results.results_dict["metrics/mAP50(B)"]),
    "mAP50-95": float(results.results_dict["metrics/mAP50-95(B)"]),
    "Precision": float(results.results_dict["metrics/precision(B)"]),
    "Recall": float(results.results_dict["metrics/recall(B)"]),
    "Dataset Size": 11,  # Change this if different
    "epochs": 50,  # Change this if different

    "Notes": "Image From Internet"
}

# Convert to DataFrame
df = pd.DataFrame([summary])

# Save/append to log file
log_path = "logs/training_log.csv"

# If file exists, append. Else, create.
if os.path.exists(log_path):
    df.to_csv(log_path, mode='a', header=False, index=False)
else:
    df.to_csv(log_path, index=False)

print("✅ Log saved to:", log_path)
```
