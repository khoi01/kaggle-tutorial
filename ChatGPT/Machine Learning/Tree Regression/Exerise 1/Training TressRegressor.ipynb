```python
# Re-import necessary libraries after execution state reset
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor, plot_tree
from sklearn.preprocessing import LabelEncoder

import pandas as pd

#import data from .csv file
data=pd.read_csv('camera_rental_data.csv')
```

```python
print(data)
```

```python
le = LabelEncoder()
data['Camera_Model_Numeric'] = le.fit_transform(data['Camera_Model'])
data.info()
```

```python
#sort column
data = data.sort_values(by=["Customer_Name", "Camera_Model", "Days_Rented"])
```

```python
data
```

```python

#features
X = data[['Camera_Model_Numeric','Days_Rented']]
y = data['Days_Rented']

#Train TreeRegressor Model
tree_reg = DecisionTreeRegressor(max_depth=3, random_state=42)
tree_reg.fit(X, y)
```

```python
# Train a simple Decision Tree Regression model
tree_reg = DecisionTreeRegressor(max_depth=3, random_state=42)
tree_reg.fit(X, y)

# Plot the decision tree
plt.figure(figsize=(12, 6))
plot_tree(tree_reg, feature_names=['Camera_Model_Numeric', 'Days_Rented'], filled=True, rounded=True)
plt.show()
```
