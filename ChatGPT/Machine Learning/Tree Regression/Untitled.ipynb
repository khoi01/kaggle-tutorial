```python
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor,plot_tree
import pandas as pd
```

```python
# Create a simple dataset for second-hand Sony camera prices
data = pd.DataFrame({
    'Age_Years': [1, 2, 3, 4, 5],  # Age of the camera in years
    'Shutter_Count': [5000, 15000, 30000, 50000, 80000],  # Number of shutter actuations
    'Price': [4000, 3500, 3000, 2500, 2000]  # Second-hand price in currency (e.g., MYR/USD)
})

# Features and target
X = data[['Age_Years', 'Shutter_Count']]
y = data['Price']
```

```python
# Train a simple Decision Tree Regression model
tree_reg = DecisionTreeRegressor(max_depth=3, random_state=42)
tree_reg.fit(X, y)
```

```python
# Plot the decision tree
plt.figure(figsize=(12, 6))
plot_tree(tree_reg, feature_names=['Age_Years', 'Shutter_Count'], filled=True, rounded=True)
plt.show()
```

```python
# Example: Predict price for a 2.5-year-old camera with 20000 shutter count
new_camera = [[2.5, 20000]]
predicted_price = tree_reg.predict(new_camera)

print(f"Predicted Price: MYR {predicted_price[0]:,.2f}")
```
