```python
# Import necessary libraries
import pandas as pd
from sklearn.tree import DecisionTreeRegressor
from sklearn.preprocessing import LabelEncoder

# Load the dataset
file_path = "camera_rental_data.csv"
rental_data = pd.read_csv(file_path)

# Encode categorical variables (Customer Name & Camera Model)
label_encoder_customer = LabelEncoder()
label_encoder_camera = LabelEncoder()

rental_data['Customer_Name'] = label_encoder_customer.fit_transform(rental_data['Customer_Name'])
rental_data['Camera_Model'] = label_encoder_camera.fit_transform(rental_data['Camera_Model'])

# Define features (X) and target (y)
X = rental_data[['Customer_Name', 'Camera_Model']]
y = rental_data[['Days_Rented', 'Total_Cost']]  # Multiple target variables

# Train a DecisionTreeRegressor model
tree_reg = DecisionTreeRegressor(random_state=43)  # Setting random_state for reproducibility
tree_reg.fit(X, y)
```
