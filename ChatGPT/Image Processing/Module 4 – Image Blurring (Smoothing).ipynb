```python
# ====================================
# Module 4: Image Blurring (Smoothing)
# ====================================

# Blurring is a technique to reduce image noise and details.
# It works by averaging pixel values with their neighbors.
# This helps to remove small textures or irrelevant details before thresholding or edge detection.

# Why blur an image?
# - To smooth out the background
# - To reduce noise before detecting objects (like bees)
# - To improve thresholding and contour detection

# Common types of blurring:
# 1. Average Blur: simple and fast
# 2. Gaussian Blur: weighted average, looks more natural
# 3. Median Blur: good for removing salt-and-pepper noise
# 4. Bilateral Filter: reduces noise while keeping edges sharp (used for face or object segmentation)

# Important:
# - Blurring affects the whole image, including the object (e.g., a bee).
# - Use small blur sizes to reduce background noise without removing important object details.

# After blurring, apply thresholding or edge detection for better results.
```

```python
import cv2
import matplotlib.pyplot as plt

# Load grayscale image
img = cv2.imread('kelulut1.jpg', cv2.IMREAD_GRAYSCALE)

# 1. Apply different blurring methods
blur = cv2.blur(img, (5,5))
gaussian = cv2.GaussianBlur(img, (5,5), 0)
median = cv2.medianBlur(img, 5)
bilateral = cv2.bilateralFilter(img, 9, 75, 75)

# 2. Display all blurring results
titles = ['Original', 'Average Blur', 'Gaussian Blur', 'Median Blur', 'Bilateral Filter']
images = [img, blur, gaussian, median, bilateral]

plt.figure(figsize=(15,8))
for i in range(5):
    plt.subplot(2, 3, i+1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Load image in grayscale
img = cv2.imread("kelulut1.jpg", cv2.IMREAD_GRAYSCALE)

# Apply strong blur for testing
blurred = cv2.GaussianBlur(img, (25,25), 0)

# Calculate Laplacian Variance (sharpness measure)
original_var = cv2.Laplacian(img, cv2.CV_64F).var()
blurred_var = cv2.Laplacian(blurred, cv2.CV_64F).var()

print("Original Sharpness (Laplacian Variance):", original_var)
print("Blurred Sharpness (Laplacian Variance):", blurred_var)

# Visual comparison
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title(f"Original (Var: {original_var:.2f})")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(blurred, cmap='gray')
plt.title(f"Blurred (Var: {blurred_var:.2f})")
plt.axis('off')
plt.show()
```
