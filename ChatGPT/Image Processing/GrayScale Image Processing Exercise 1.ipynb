```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('khoi.jpg') #load image
img_rgb = cv2.cvtColor(img,cv2.COLOR_BGR2RGB) # convert to rgb for mathplotlib purpose
gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY) # convert to gray


#Display original and grayscale image side by side
plt.figure(figsize=(12,5))

plt.subplot(1,3,1)
plt.imshow(img_rgb)
plt.title("Original RGB Image")

plt.subplot(1, 2, 2)
plt.imshow(gray,cmap='gray')
plt.title("Gray Scale Image")

plt.show()

#isplay histogram of grayscale image
plt.figure(figsize=(6,4))
plt.hist(gray.ravel(),bins=256,range=[0,255])
plt.title("Histogram of Grayscale Image")
plt.xlabel("Pixel Intensity")
plt.ylabel("Count")
plt.show()

#cv2.threshold() is used to convert a grayscale image into a binary image (black and white only).
#If pixel intensity > 127 → set to 0 (black)
#Else → set to 255 (white)
_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

# 7. Show binary image
plt.imshow(binary, cmap='gray')
plt.title("Binary Threshold Result")
plt.axis("off")
plt.show()

#cv2.THRESH_BINARY_INV It’s the inverse (the opposite):
_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY_INV)

# 7. Show binary image
plt.imshow(binary, cmap='gray')
plt.title("Binary Threshold Result")
plt.axis("off")
plt.show()

#isplay histogram of binary image
plt.figure(figsize=(6,4))
plt.hist(binary.ravel(),bins=256,range=[0,255])
plt.title("Histogram of Grayscale Image")
plt.xlabel("Pixel Intensity")
plt.ylabel("Count")
plt.show()


```
