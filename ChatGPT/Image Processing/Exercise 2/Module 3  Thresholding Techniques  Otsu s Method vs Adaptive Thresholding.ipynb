```python
import cv2
import matplotlib.pyplot as plt

# Load grayscale image
img = cv2.imread('khoi.jpg', cv2.IMREAD_GRAYSCALE)

# Display original image
plt.figure(figsize=(5,5))
plt.imshow(img, cmap='gray')
plt.title("Original Grayscale Image")
plt.axis('off')
plt.show()

_, global_thresh = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)
plt.imshow(global_thresh, cmap='gray')
plt.title("Global Threshold (127)")
plt.axis('off')
plt.show()

# Otsu finds the best threshold automatically
_, otsu_thresh = cv2.threshold(img, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

print(f"Otsu’s chosen threshold: {_}")  # Display the chosen threshold

plt.imshow(otsu_thresh, cmap='gray')
plt.title("Otsu's Thresholding (Auto)")
plt.axis('off')
plt.show()
```

```python
# Adaptive Mean Thresholding
adaptive_thresh_mean = cv2.adaptiveThreshold(img, 255,
                                             cv2.ADAPTIVE_THRESH_MEAN_C,
                                             cv2.THRESH_BINARY,
                                             11, 2)

# Adaptive Gaussian Thresholding
adaptive_thresh_gauss = cv2.adaptiveThreshold(img, 255,
                                              cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
                                              cv2.THRESH_BINARY,
                                              11, 2)

# Show both results
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(adaptive_thresh_mean, cmap='gray')
plt.title("Adaptive Threshold (Mean)")

plt.subplot(1, 2, 2)
plt.imshow(adaptive_thresh_gauss, cmap='gray')
plt.title("Adaptive Threshold (Gaussian)")

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

#load image and convert to grayscale
img = cv2.imread('cars.jpg',cv2.IMREAD_GRAYSCALE)

#apply global threshold
_,global_thresh = cv2.threshold(img,127,255,cv2.THRESH_BINARY)

#apply otsu threshold
_,otsu_thresh = cv2.threshold(img,0,255,cv2.THRESH_BINARY + cv2.THRESH_OTSU)

# Apply Adaptive Threshold (Mean)
adaptive_thresh = cv2.adaptiveThreshold(img, 255,
                                        cv2.ADAPTIVE_THRESH_MEAN_C,
                                        cv2.THRESH_BINARY,
                                        11, 2)
# Show all results
plt.figure(figsize=(12,6))
titles = ['Original', 'Global Threshold', "Otsu's Threshold", 'Adaptive Threshold']
images = [img, global_thresh, otsu_thresh, adaptive_thresh]

for i in range(4):
    plt.subplot(2,2,i+1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# 1. Load your image (leaf on soil / bee on hive)
img = cv2.imread("kelulut1.jpg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# 2. Global Thresholding (fixed value)
_, global_thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

# 3. Otsu's Thresholding (auto)
_, otsu_thresh = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)

# 4. Adaptive Thresholding (mean)
adaptive_thresh = cv2.adaptiveThreshold(gray, 255,
                                        cv2.ADAPTIVE_THRESH_MEAN_C,
                                        cv2.THRESH_BINARY,
                                        11, 3)

# 5. Inverted Adaptive (if object is dark)
adaptive_inv = cv2.adaptiveThreshold(gray, 255,
                                     cv2.ADAPTIVE_THRESH_MEAN_C,
                                     cv2.THRESH_BINARY_INV,
                                     11, 2)

# 6. Display results
plt.figure(figsize=(12, 8))

titles = ['Original Grayscale', 'Global (127)', "Otsu's Method", 
          'Adaptive Mean', 'Adaptive Mean Inverse']
images = [gray, global_thresh, otsu_thresh, adaptive_thresh, adaptive_inv]

for i in range(5):
    plt.subplot(2, 3, i+1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()
```
