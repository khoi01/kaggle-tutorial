```python
# ==============================================
# Module 6: Morphological Operations - Concept
# ==============================================

# Morphological operations are techniques used to clean up binary images (black and white).
# They are especially useful after thresholding or edge detection.

# Why use them?
# - To remove small noise dots (false positives)
# - To close small holes inside objects (e.g., broken bee shape)
# - To improve object shapes before contour detection

# Morphology uses a small matrix called a "kernel" to process the image pixel by pixel.

# Common operations:
# 1. Erosion: erodes away white pixels (shrinks object)
# 2. Dilation: adds white pixels (grows object)
# 3. Opening: erosion followed by dilation (removes small noise)
# 4. Closing: dilation followed by erosion (fills small holes)

# Use on binary images where the object is white and background is black.
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

# Load image and convert to binary (thresholding)
img = cv2.imread('kelulut1.jpg', cv2.IMREAD_GRAYSCALE)
_, binary = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)

# Define a kernel (structuring element)
kernel = np.ones((5,5), np.uint8)  # You can try (3,3), (7,7), etc.

# Apply morphological operations
erosion = cv2.erode(binary, kernel, iterations=1)
dilation = cv2.dilate(binary, kernel, iterations=1)
opening = cv2.morphologyEx(binary, cv2.MORPH_OPEN, kernel)
closing = cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernel)

# Show results
titles = ['Original Binary', 'Erosion', 'Dilation', 'Opening', 'Closing']
images = [binary, erosion, dilation, opening, closing]

plt.figure(figsize=(15, 8))
for i in range(5):
    plt.subplot(2, 3, i+1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()
```
