```python
# ====================================
# Module 5: Edge Detection
# ====================================

# Edge detection is used to find the outlines or boundaries of objects in an image.
# It works by looking for areas where pixel values change quickly — which usually means an edge.

# Why is edge detection important?
# - It highlights object shapes (like bees, leaves, or tools)
# - It helps in counting, contour detection, and feature extraction
# - It reduces image complexity while keeping key structures

# Common edge detection techniques:
# 1. Sobel Filter: detects edges in horizontal or vertical direction
# 2. Laplacian Filter: detects edges in all directions
# 3. Canny Edge Detection: the most powerful and accurate method

# Notes:
# - Apply edge detection on a grayscale image (not color)
# - Blurring before edge detection helps reduce false edges caused by noise
# - Edges appear as white lines on a black background (binary image)

# After detecting edges, you can:
# - Count objects
# - Find contours
# - Extract shapes or outlines for classification
```

```python
import cv2
import matplotlib.pyplot as plt

# 1. Load the image in grayscale
img = cv2.imread('kelulut1.jpg', cv2.IMREAD_GRAYSCALE)

# 2. Optional: Blur the image to reduce noise before edge detection
blurred = cv2.GaussianBlur(img, (5, 5), 0)

# 3. Sobel Edge Detection (Horizontal and Vertical)
sobelx = cv2.Sobel(blurred, cv2.CV_64F, 1, 0, ksize=3)  # Horizontal edges
sobely = cv2.Sobel(blurred, cv2.CV_64F, 0, 1, ksize=3)  # Vertical edges
sobel_combined = cv2.magnitude(sobelx, sobely)

# 4. Laplacian Edge Detection (All directions)
laplacian = cv2.Laplacian(blurred, cv2.CV_64F)

# 5. Canny Edge Detection (cleanest)
canny = cv2.Canny(blurred, 100, 200)  # Adjust values for stronger/weaker edges

# 6. Show the results
plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Grayscale')
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(sobelx, cmap='gray')
plt.title('Sobel - Horizontal')
plt.axis('off')

plt.subplot(2, 3, 3)
plt.imshow(sobely, cmap='gray')
plt.title('Sobel - Vertical')
plt.axis('off')

plt.subplot(2, 3, 4)
plt.imshow(sobel_combined, cmap='gray')
plt.title('Sobel - Combined')
plt.axis('off')

plt.subplot(2, 3, 5)
plt.imshow(laplacian, cmap='gray')
plt.title('Laplacian')
plt.axis('off')

plt.subplot(2, 3, 6)
plt.imshow(canny, cmap='gray')
plt.title('Canny Edge Detection')
plt.axis('off')

plt.tight_layout()
plt.show()
```
