```python
# Define expected data types
expected_types = {
    'ID': int,
    'Age': int,
    'Salary': float,
    'Join_Date': 'datetime64[ns]',
    'Active': bool
}

# Function to count wrong datatype entries
wrong_type_counts = {}

for col, expected_type in expected_types.items():
    if expected_type == 'datetime64[ns]':
        # Check if the column can be converted to datetime
        wrong_type_counts[col] = data[col].apply(lambda x: pd.to_datetime(x, errors='coerce') is pd.NaT).sum()
    else:
        # Check wrong datatype entries
        wrong_type_counts[col] = data[col].apply(lambda x: not isinstance(x, expected_type) and pd.notna(x)).sum()

# Display results
print("Number of wrong datatype entries per column:")
print(wrong_type_counts)
```
