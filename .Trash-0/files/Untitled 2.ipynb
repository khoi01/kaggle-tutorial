```python
from pathlib import Path
import pandas as pd
from IPython.display import display

# Setup paths
base_path = Path.cwd()
dataset_yaml = base_path / "dataset.yaml"
pretrained_model = base_path / "yolov8n.pt"
images_train = base_path / "datasets/v1/images/train"
images_val = base_path / "datasets/v1/images/val"
labels_train = base_path / "datasets/v1/labels/train"
labels_val = base_path / "datasets/v1/labels/val"

# Checklist
checklist = {
    "✅ dataset.yaml exists": dataset_yaml.exists(),
    "✅ yolov8n.pt exists": pretrained_model.exists(),
    "✅ images/train folder exists": images_train.exists(),
    "✅ images/val folder exists": images_val.exists(),
    "✅ labels/train folder exists": labels_train.exists(),
    "✅ labels/val folder exists": labels_val.exists(),
}

def count_matches(image_dir, label_dir, ext=".JPG"):
    img_files = {f.stem for f in image_dir.glob(f"*{ext}")}
    lbl_files = {f.stem for f in label_dir.glob("*.txt")}
    return len(img_files & lbl_files), len(img_files), len(lbl_files)

train_match, train_imgs, train_lbls = count_matches(images_train, labels_train)
val_match, val_imgs, val_lbls = count_matches(images_val, labels_val)

# Display results
results_df = pd.DataFrame(list(checklist.items()), columns=["Check", "Status"])
summary_df = pd.DataFrame([
    ["Train", train_imgs, train_lbls, train_match],
    ["Validation", val_imgs, val_lbls, val_match]
], columns=["Split", "Images", "Labels", "Matched Pairs"])

display(results_df)
display(summary_df)
```
