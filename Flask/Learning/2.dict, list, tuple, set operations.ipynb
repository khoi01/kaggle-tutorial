```python
#List

my_list = [1,2,3]
my_list.append(4)
my_list.insert(2,10) #position 2 and override value.
my_list.remove(10)
my_list[0] = 10
print(my_list)
```

```python
# tuple

my_tuple = (1,2)
print(my_tuple[1])
```

```python
#Set

my_set = {1,2,3,4}
my_set.add(5)
my_set.remove(5)
print(my_set)
```

```python
#Dictionary

my_dic = {"Name":"khoi",
          "Age":34
         }
my_dic['Age'] = 35
my_dic['Blood'] = 'O'
print(my_dic)
print(my_dic['Name'])
```
