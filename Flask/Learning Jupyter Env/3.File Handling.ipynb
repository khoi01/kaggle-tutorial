```python
file = open("note.txt", "r")
```

```python
#read entire failed
content = file.read() 
print(content)
```

```python
#read 1 row.
content = file.readline()
print(content)
```

```python
#read multiple row
content = file.readlines()     # Read all lines into a list
print(content)
```

```python
file.close()
```

```python
#writing and its will override whole text on the file. (remove previous text)
file = open("note.txt","w")
file.write("edit v1 \n")
file.close()
```

```python
#append text on the file
file = open("note.txt","a")
file.write("add new text v1")
file.close()
```

```python
#best practice
with open("note.txt","r") as file:
    context = file.read()
    print(context)
```
