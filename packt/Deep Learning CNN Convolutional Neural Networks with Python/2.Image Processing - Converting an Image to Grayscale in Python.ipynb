```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
im = plt.imread('khoi.jpg')
```

```python
#display image
plt.imshow(im)
```

```python
#identify height,width and channel (RGB)
im.shape
#if the channel more then 3 then remove 1 channel
im2 = im[:,:,:3]
im2.shape
```

```python
R = im2[:,:,0]
G = im2[:,:,1]
B = im2[:,:,2]
```

```python
imGray = 0.333*R + 0.333*G + 0.333*B
plt.imshow(imGray,cmap='gray')
```

```python
imGray = 0.298*R + 0.5870*G + 0.1140*B
plt.imshow(imGray,cmap='gray')
```

```python
#open opencv2 grey

import cv2
imGrayCV2 = cv2.cvtColor(im2,cv2.COLOR_RGB2GRAY)

fig,(ax1,ax2) = plt.subplots(1,2,figsize=(10,15))
ax1.imshow(imGray,cmap='gray')
ax2.imshow(imGrayCV2,cmap='gray')
```
