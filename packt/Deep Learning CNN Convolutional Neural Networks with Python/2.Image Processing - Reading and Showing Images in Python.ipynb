```python
import matplotlib.pyplot as plt

R = im2[:,:,0] #RED Channel
G = im2[:,:,1] #GREEN Channel
B = im2[:,:,2] #BLUE Channel

#plt.subplots(rows, cols, figsize=(width, height))
# Breaking Down the Parameters:
# 1 → Number of rows (1 row)
# 3 → Number of columns (3 columns)
# → This means we are creating 3 subplots in a single row.
# figsize=(15, 5) → Figure size in inches
# 15 → Width of the entire figure (in inches)
# 5 → Height of the entire figure (in inches)


fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 5))

ax1.imshow(R, cmap='gray')
ax1.set_title('Red Channel')

ax2.imshow(G, cmap='gray')
ax2.set_title('Green Channel')

ax3.imshow(B, cmap='gray')
ax3.set_title('Blue Channel')

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

#fig → Represents the overall figure (canvas where all subplots are placed).
#plt.subplots(1, 3)	  Creates 1 row and 3 columns of subplots
#fig The overall figure (canvas)
#(ax1, ax2, ax3)	 Individual subplot axes
#figsize=(15,5)	Figure width = 15 inches, height = 5 inches

# Create dummy grayscale images
R = np.random.rand(100, 100)  # Red channel (grayscale)
G = np.random.rand(100, 100)  # Green channel (grayscale)
B = np.random.rand(100, 100)  # Blue channel (grayscale)

# Create a figure with 1 row and 3 columns
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 5))

# Plot each channel
ax1.imshow(R, cmap='gray')  
ax1.set_title("Red Channel")

ax2.imshow(G, cmap='gray')  
ax2.set_title("Green Channel")

ax3.imshow(B, cmap='gray')  
ax3.set_title("Blue Channel")

plt.show()
```
