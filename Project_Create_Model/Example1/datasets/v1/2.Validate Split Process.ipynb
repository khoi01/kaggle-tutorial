```python
from pathlib import Path

img_dir = Path("images/train")
lbl_dir = Path("labels/train")

mismatched = []

for img_file in img_dir.glob("*.JPG"):
    lbl_file = lbl_dir / (img_file.stem + ".txt")
    if not lbl_file.exists():
        mismatched.append(img_file.name)

if mismatched:
    print("❌ Mismatched image files with no label:")
    for m in mismatched:
        print("-", m)
else:
    print("✅ All training images have matching labels.")
```

```python
from pathlib import Path

# Define folders
splits = ['train', 'val']
base_img_path = Path("images")
base_lbl_path = Path("labels")

for split in splits:
    print(f"\n🔍 Checking '{split}' split:")
    img_dir = base_img_path / split
    lbl_dir = base_lbl_path / split
    mismatched = []

    for img_file in img_dir.glob("*.JPG"):
        lbl_file = lbl_dir / (img_file.stem + ".txt")
        if not lbl_file.exists():
            mismatched.append(img_file.name)

    if mismatched:
        print(f"❌ Found {len(mismatched)} image(s) in '{split}' with no matching label:")
        for m in mismatched:
            print(" -", m)
    else:
        print(f"✅ All {split} images have matching labels.")
```
